```python
import numpy as np
```

```python
a = np.array([[1,2],[3,4]])
print("A numpy array of size 2 x 2: \n", a)
```

```python
b = np.random.randint(1,20,size=(3,4))
print("A random numpy array of size 3 x 4:\n", b)
```

```python
c = np.ones((3,4))
d = np.zeros((2,5))

print("A numpy array of ones:\n", c)
print("A numpy array of zeros:\n", d)
```

```python
e = np.random.randint(1,20,size=(6,3))
eTransposed = e.T

print("Transpose of a (6x3) random numpy array:\n", eTransposed)
```

```python
n=3
p=4
m=5

f = np.array([n,p])
g = np.array([p,m])
print("Array 1(n x p):\n", f)
print("Array 2(p x m):\n", g)

dotproduct = f @ g
print("The dot product of f(nxp) and g(pxm):\n",dotproduct)
```
